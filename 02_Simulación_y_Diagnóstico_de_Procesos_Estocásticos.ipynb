## 1. Introducción
En la teoría de procesos estocásticos, los conceptos de **estacionariedad** y **ergodicidad** son pilares fundamentales que nos permiten simplificar el análisis de sistemas complejos en el tiempo. Mientras que la estacionariedad en sentido amplio (WSS) garantiza que las propiedades estadísticas de primer y segundo orden (media y autocovarianza) son invariantes ante desplazamientos temporales, la ergodicidad nos otorga una ventaja matemática y práctica crucial: estimar los parámetros de todo un ensamble teórico utilizando una **única trayectoria** observada en un horizonte temporal lo suficientemente largo.

En esta práctica, utilizaremos simulaciones computacionales en Python para contrastar procesos teóricos ideales con datos simulados, desarrollando una sólida intuición visual y estadística sobre estos conceptos abstractos.
